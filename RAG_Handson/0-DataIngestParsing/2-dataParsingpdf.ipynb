```python
from langchain_community.document_loaders import(
    PyPDFLoader,
    PyMuPDFLoader
)


```

```python
print("PyPDFLoader")

try:
    pypdf_loader = PyPDFLoader("data/pdf/Rahul.pdf")
    pypdf_doc = pypdf_loader.load()
    print(pypdf_doc)
except Exception as e:
    print(f"Error : {e}")
```

```python
print("PyMuPDFLoader")

try:
    pymupdf_loader = PyMuPDFLoader("data/pdf/Rahul.pdf")
    pymupdf_docs = pymupdf_loader.load()
    print(f"Loaded {len(pymupdf_docs)} pages")
    print(f"Include detailed metadata")
    print(pymupdf_docs)
except Exception as e:
    print(f"Error: {e}")
```
